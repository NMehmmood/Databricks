```sql
--classical views 
CREATE VIEW newdevice_view AS SELECT * FROM devices WHERE year > 2010;
```

```sql
--Temporary view
CREATE TEMP VIEW old_devices AS SELECT * FROM devices WHERE year < 2011;
```

```sql
SELECT * FROM old_devices;
```

```sql
SHOW TABLES
```

```sql
---Global Temporary View
CREATE OR REPLACE GLOBAL TEMP VIEW latest_devices 
AS SELECT * FROM devices where year > 2020;
```

```sql
SELECT * FROM global_temp.latest_devices;
```
